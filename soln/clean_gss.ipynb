## Loading

Downloaded February 2021, from https://gssdataexplorer.norc.org/projects/52787

```python
import pandas as pd
import numpy as np
```

```python
from statadict import parse_stata_dict
import gzip

def read_gss(dict_file='GSS.dct', data_file='GSS.dat.gz'):
    stata_dict = parse_stata_dict(dict_file)
    fp = gzip.open(data_file)
    gss = pd.read_fwf(fp,
                      names=stata_dict.names,
                      colspecs=stata_dict.colspecs)
    return gss
```

```python
gss = read_gss()
gss.shape
```

```python
gss['AGE'].replace([98, 99], np.nan, inplace=True)
gss['EDUC'].replace([98, 99], np.nan, inplace=True)
gss['GUNLAW'].replace([8, 9, 0], np.nan, inplace=True)
gss['GRASS'].replace([8, 9, 0], np.nan, inplace=True)
gss['REALINC'].replace([0], np.nan, inplace=True)
```

```python
subset_columns = ['YEAR', 'ID_', 'AGE', 'EDUC', 'SEX', 
                  'GUNLAW', 'GRASS', 'REALINC']

subset = gss[subset_columns]
subset.head()
```

```python
!rm gss_eda.hdf
```

```python
subset.to_hdf('gss_eda.hdf', 'gss', complevel=3)
```
